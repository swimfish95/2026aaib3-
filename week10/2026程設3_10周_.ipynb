```cpp
// week10-1a.cpp 學習計畫 Math 第1題 (暴力 for 迴圈,沒用到數學)
// LeetCode 1523. Count Odd Numbers in an Interval Range
class Solution {
public:
    int countOdds(int low, int high) {
        int ans = 0;
        for (int i=low; i<=high; i++) { // 迴圈很慢
            if (i%2==1) ans++; // 有幾個「奇數」
        }
        return ans;
    }
};
```

```cpp
// week10-1b.cpp 學習計畫 Math 第1題 (用數學)
// LeetCode 1523. Count Odd Numbers in an Interval Range
class Solution {
public:
    int countOdds(int low, int high) {
        int ans = (high - low) / 2;
        if (low%2==1 || high%2==1) ans++;
        return ans;
    }
};
```

```cpp
// week10-2.cpp SOIT106_Advance_002
#include <iostream>
using namespace std;
int isPrime(int N) {
    if (N < 2) return 0;               // 0 和 1 不是質數
    // 除了 1 跟 N 本身，都不整除
    for (int i=2; i*i<=N; i++) {       // 只需檢查到 √N
        if ( N%i==0 ) return 0;
    }
    return 1;
}
int main()
{
    int a, b;
    cin >> a >> b;
    int ans = 0;
    for (int i=a; i<=b; i++) {
        if ( isPrime(i) ) ans++;
    }
    cout << ans << "\n";
    return 0;
}
```

```cpp
// week10-3.cpp 學習計畫 Math 第2題
// LeetCode 1491. Average Salary Excluding the Minimum and Maximum Salary
// 把最高薪、最低薪，扣掉。剩下的平均
class Solution {
public:
    double average(vector<int>& salary) {
        double total = 0; // 總薪水
        int M = salary[0], m = salary[0];
        for (int now : salary) { // 用迴圈，逐一查看薪水
            if (now > M) M = now;
            if (now < m) m = now;
            total += now; // 把薪水加進 total
        }
        return (total - M - m) / (salary.size() - 2);
        // return total / salary.size(); 錯錯錯，要扣掉 最高薪、最低薪
    }
};
```

```cpp
class Solution {
public:
    string multiply(string num1, string num2) {
        int m = num1.size(), n = num2.size();
        vector<int> pos(m + n, 0);
        
        for (int i = m - 1; i >= 0; i--) {
            for (int j = n - 1; j >= 0; j--) {
                int mul = (num1[i] - '0') * (num2[j] - '0');
                int p1 = i + j, p2 = i + j + 1;
                int sum = mul + pos[p2];
                
                pos[p2] = sum % 10;
                pos[p1] += sum / 10;
            }
        }
        
        string result;
        for (int p : pos) {
            if (!(result.empty() && p == 0)) {
                result += to_string(p);
            }
        }
        
        return result.empty() ? "0" : result;
    }
};
```
