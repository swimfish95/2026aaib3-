```cpp
#include <iostream>
using namespace std;

int MYPOWER(int a, int b) {
    int ans = 1;
    for (int i=0; i<b; i++) {
        ans = ans * a;
    }
    return ans;
}

int main(){
    int a,b;cin>>a>>b;
    cout<<MYPOWER(a,b);
    return 0;
}
```

```cpp
class Solution {
public:
    int calPoints(vector<string>& operations) {
        vector<int> record;
        for (const string& op : operations) {
            if (op == "+") {
                int n = record.size();
                record.push_back(record[n - 1] + record[n - 2]);
            } else if (op == "D") {
                record.push_back(2 * record.back());
            } else if (op == "C") {
                record.pop_back();
            } else {
                record.push_back(stoi(op));
            }
        }
        int sum = 0;
        for (int score : record) sum += score;
        return sum;
    }
};
```

```cpp
//week07-3
class Solution {
public:
    bool isRobotBounded(string instructions) {
        int x = 0, y = 0, d = 0; // d方向 0:北 1:東 2:南 3:西
        // 重複4次
        instructions = instructions + instructions + instructions + instructions;
        for (char c : instructions) {
            if (c == 'R') d = (d + 1) % 4; // Right 右轉
            if (c == 'L') d = (d + 3) % 4; // Left 左轉 (1個左轉=3個右轉)
            if (c == 'G') { // Go 往前走1格
                if (d == 0) y++;
                if (d == 1) x++;
                if (d == 2) y--;
                if (d == 3) x--;
            }
        }
        if (x == 0 && y == 0) return true;  // 成功
        else return false;                  // 沒有回到原來的位置,失敗
    }
};

```

```cpp
class Solution {
public:
    string tictactoe(vector<vector<int>>& moves) {
        int a[3][3] = {0};                       // 3x3 格子,先放0
        string result[3] = {"Draw", "A", "B"};   // 結果的字母
        int who = 1;                             // 現在輪到誰下棋 1:A, 2:B
        for (vector<int> move : moves) {
            int i = move[0], j = move[1];
            a[i][j] = who;                       // 現在輪到誰就放誰
            // 三橫
            if (a[0][0]==who && a[0][1]==who && a[0][2]==who) return result[who];
            if (a[1][0]==who && a[1][1]==who && a[1][2]==who) return result[who];
            if (a[2][0]==who && a[2][1]==who && a[2][2]==who) return result[who];
            // 三直
            if (a[0][0]==who && a[1][0]==who && a[2][0]==who) return result[who];
            if (a[0][1]==who && a[1][1]==who && a[2][1]==who) return result[who];
            if (a[0][2]==who && a[1][2]==who && a[2][2]==who) return result[who];
            // 兩條斜線
            if (a[0][0]==who && a[1][1]==who && a[2][2]==who) return result[who];
            if (a[0][2]==who && a[1][1]==who && a[2][0]==who) return result[who];

            if (who == 1) who = 2;
            else who = 1;
        }
        if (moves.size() == 9) return "Draw";   // 9格都下滿,沒人贏
        return "Pending";                        // 還有空格,遊戲未結束
    }
};    
```
