```cpp
class Solution {
public:
    ListNode* reverseList(ListNode* head) {
        // Base case：空串列或只剩最後一個節點
        if (head == nullptr || head->next == nullptr) {
            return head;
        }
        
        // 遞迴反轉後面的串列，newHead 永遠會是原串列的最後一個節點
        ListNode* newHead = reverseList(head->next);
        
        head->next->next = head; // 讓下一個節點指回自己
        head->next = nullptr;    // 斷開舊連結，防止成環
        
        return newHead;
    }
};
```

```cpp
#include <iostream>
#include <vector>
using namespace std;
int main()
{
    int N;
    cin >> N;
    vector<int> a(N);
    for (int i=0; i<N; i++) {
        cin >> a[i];
    }
    for (int i=N-1; i>=0; i--) {
        if (a[i]%2==1) cout << a[i] << " ";
    }
}
```

```cpp
class Solution {
public:
    ListNode* mergeTwoLists(ListNode* list1, ListNode* list2) {
        ListNode dummy(0);       // 虛擬頭節點（宣告在 stack 上，不需手動 delete）
        ListNode* tail = &dummy; // tail 永遠指向目前合併結果的最後一個節點
        
        while (list1 != nullptr && list2 != nullptr) {
            if (list1->val <= list2->val) {
                tail->next = list1;
                list1 = list1->next;
            } else {
                tail->next = list2;
                list2 = list2->next;
            }
            tail = tail->next;   // 推進 tail
        }
        
        // 其中一條已經走完，直接把另一條剩下的部分整串接上
        tail->next = (list1 != nullptr) ? list1 : list2;
        
        return dummy.next;
    }
};
```
