```cpp
// week06-1.cpp UVA10696
#include <iostream>
int f91(int N) {
    if (N<=100) return f91(f91(N+11));
    else return N-10;
}
using namespace std;
int main()
{
    int N;
    while ( cin >> N ) {
        if (N==0) break;
        printf("f91(%d) = %d\n", N, f91(N) );
    }
}
```

```cpp
// next week exam
// SOIT108_Advance_014B
#include <iostream>
using namespace std;
int main()
{
    int N;
    cin >> N;
    int ten = 1;
    while (N>0) {
        int now = N%10 * ten;
        ten = ten * 10;
        N = N / 10;
        cout << now << " ";
    }
}
```

```cpp
class Solution {
public:
    bool judgeCircle(string moves) {
        int x = 0, y = 0;
        for ( char c : moves ) {
            if (c=='U') y++;
            if (c=='D') y--;
            if (c=='L') x--;
            if (c=='R') x++;
        }
        return x == 0 && y == 0;
    }
};
```
